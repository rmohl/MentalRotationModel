```python
import numpy as np
import math
from hrr import hrr
from scipy.spatial.transform import Rotation as rotate
```

```python
def cosine_similarity(vec_a, vec_b):
    return np.dot(vec_a, vec_b) / np.linalg.norm(vec_a) * np.linalg.norm(vec_b)
```

```python
n = 16  # number of neurons
```

```python
preferred_directions = np.linspace(0, 2 * np.pi, n, endpoint=False)  # create direction angles (in radians)
print("Radians: " + str(preferred_directions))
print("Degrees: " + str(np.degrees(preferred_directions)))
```

```python
direction_vectors = np.array([
    [np.sin(theta), np.cos(theta), 0]  # z-component is zero, for now
    for theta in preferred_directions
])
print(direction_vectors)
```

```python
'''
how much do neurons fire for a direction? 
- 1 for same
- 0 for complete opposite
- how do we do in between? --> gaussian or von Mises function

important to note that in npv paper the population vectors started with small magitude and increased as the turn completed (increase looks like y = logx graph)
'''

θ_target = np.radians(90)
target_vector = np.array([np.sin(θ_target), np.cos(θ_target), 0])
θ_current = np.radians(0)
current_vector = np.array([np.sin(θ_current), np.cos(θ_current), 0])

print(target_vector, current_vector)
print(cosine_similarity(current_vector, target_vector))

step_size = 5
rotation_axis = np.array([0, 0, 1])     # z-axis for now
direction = -1                          # -1 = left, 1 = right
r = rotate.from_rotvec([0, 0, np.radians(direction * step_size)]) # quaternion representing 5 deg rotation around z-axis

### LOOP #### 

while cosine_similarity(current_vector, target_vector) < 0.95:     # checking cosine similarity
    print(cosine_similarity(current_vector, target_vector))

    # re-adjust axis of rotation

    # calculate population vector
    firing_rates = [0.5 * (math.cos(θ_current - θ_i) + 1) for θ_i in preferred_directions]  # use gaussian or von Mises function rather than cos() for more biologically accurate firing rates
    print(firing_rates)
    population_vector = sum(firing_rates[i] * direction_vectors[i] for i in range(n))
    print(population_vector)

    # update current vector
    current_vector = r.apply(current_vector)
    print(current_vector)

    θ_current = θ_current + np.radians(step_size)

# '''
# perform rotation using rotation matrix or quaternions?
# - chatgpt suggested quaternions (better for multiple small rotations)
# - import from scipy
# '''
print("hello")
```
