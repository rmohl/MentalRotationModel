```python
import pandas as pd

# Jost + Jonsen data
 
# with open("JostJonsenDataset.csv", "r") as old_file:

#     with open("JostJonsen_cleaned.csv", "w") as new_file:

#         for line in old_file:

#             line = line.strip()

#             if line[0] == "\"":
#                 line = line[1:]
#             if line[-1] == "\"":
#                 line = line[:-1]

#             line = line.replace("\"\"", "\"")
#             line = line.replace(";", ",")

#             new_file.write(line + "\n")
# note: col 5, "correctAnswer" was manually added!

# Ganis + Kievet data

df_demographics = pd.read_excel("GanisKievetData/demographics.xlsx")

with open("GanisKievet_cleaned.csv", "w") as new_file:

    new_file.write("subject_num,age,gender,trial_num,condition,angle,time,response,expected_answer,correct\n")

    for index_i, col_i in df_demographics.iterrows():

        if pd.isna(col_i["Subject"]):
            break

        # get demographic data
        subject_num = col_i["Subject"]
        age = col_i["Age"]
        gender = col_i["Gender"]

        # get trial data
        df_subject = pd.read_excel("GanisKievetData/sub"+str(subject_num)+".xlsx", skiprows=3)

        for index_j, col_j in df_subject.iterrows():

            if pd.isna(col_j["Trial"]):
                break

            trial_num = col_j["Trial"]
            condition = col_j["Condition"]
            time = col_j["Time"]
            response = col_j["response"]
            expected_answer = col_j["correct"]
            angle = col_j["angle"]
            correct = 1 if response == expected_answer else 0

            if pd.isna(time):
                print("(INVALID) Subject " + str(subject_num) + ", trial " + str(trial_num))
            else:
                # write data to csv
                new_file.write(str(subject_num) + "," + str(age) + "," + gender + "," + str(trial_num) + "," + condition + "," + str(angle) + "," + str(time) + "," + response + "," + expected_answer + "," + str(correct) + "\n")




```
