```python
fig = go.Figure(
    data=[
        go.Scatter3d(x=[1, 2, 3], y=[1, 2, 1], z=[1, 2, 1], mode='lines', name='Line A'),
        go.Scatter3d(x=[1, 2, 3], y=[3, 1, 3], z=[3, 1, 3], mode='lines', name='Line B')
    ],
    layout=go.Layout(
        # xaxis=dict(range=[0, 4]),
        # yaxis=dict(range=[0, 4]),
        title="Animated Multiple Lines",
        updatemenus=[dict(
            type="buttons",
            buttons=[dict(label="Play", method="animate", args=[None])]
        )]
    )
)

frames = [
    go.Frame(data=[
        go.Scatter3d(x=[1, 2, 3], y=[1, 2, 1], z=[1, 2, 1]), # Updates trace 0
        go.Scatter3d(x=[1, 2, 3], y=[3, 1, 3], z=[3, 1, 3])  # Updates trace 1
    ], name="frame1", traces=[0, 1]), # Link to initial traces
    go.Frame(data=[
        go.Scatter3d(x=[1, 2, 3], y=[2, 3, 2], z=[2, 3, 2]),
        go.Scatter3d(x=[1, 2, 3], y=[2, 2, 2], z=[2, 2, 2])
    ], name="frame2", traces=[0, 1]),
    go.Frame(data=[
        go.Scatter3d(x=[1, 2, 3], y=[3, 1, 3], z=[3, 1, 3]),
        go.Scatter3d(x=[1, 2, 3], y=[1, 3, 1], z=[1, 3, 1])
    ], name="frame3", traces=[0, 1]),
]
fig.frames = frames

fig.show()
```
