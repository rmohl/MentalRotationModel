```python
#HRRS
granularity = 16
max_geon_length = 5

granular_angles = np.linspace(0, 2 * np.pi, granularity, endpoint=False)  # create all angles (in radians)
print("Radians: " + str(granular_angles))
print("Degrees: " + str(np.degrees(granular_angles)))
```

```python
D = 512

# apply all these directions to xy plane.
# then copy all those xy plane things and add 16 to z plane(?)

xy_direction_vectors = np.array([
    [np.sin(theta), np.cos(theta), 0]  # z-component is zero, for now
    for theta in granular_angles
])
# print(xy_direction_vectors)

xyz_direction_vectors = np.zeros((granularity, granularity, 3))
print(xy_direction_vectors.shape)
print(xyz_direction_vectors.shape)
for i in range(granularity):
    z_angle = granular_angles[i]

    # update z angle in every xy coord
    for j in range(granularity):
        xy_direction_vectors[j][2] = z_angle

    xyz_direction_vectors[i] = xy_direction_vectors

print(xyz_direction_vectors.shape)

# create direction hrrs
direction_hrrs = []
for i in range(granularity):
    direction_hrrs.append([])
    for j in range(granularity):
        direction_hrrs[i].append(HRR(D))

print(len(direction_hrrs), len(direction_hrrs[0]))

# create length hrrs
length_hrrs = []

for i in range(max_geon_length):
    length_hrrs.append(HRR(D))
```
